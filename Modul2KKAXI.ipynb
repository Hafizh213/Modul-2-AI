```python
#Latihan 3

df = pd.read_csv('/content/sample_data/data_kantin.csv')
print(df.head()) # 5 baris pertama
print(df.info()) # tipe data & jumlah non-null tiap kolom
print(df.describe()) # statistik ringkas kolom numerik
print(df.shape) # jumlah (baris, kolom)
```

Analisis 3:

kolom menu dan terjual memiliki data yang kosong sehingga tidak semua baris memiliki nilai pada kedua kolom tersebut.

```python
#Latihan 4

print(df.isnull().sum()) # jumlah data kosong tiap kolom
df['terjual'] = df['terjual'].fillna(0) # isi kekosongan dengan 0
df = df.dropna(subset=['menu']) # hapus baris jika kolom menu kosong
```

Analisis 4:

Kolom terjual diisi dengan fillna() karena nilai yang hilang masih dapat diperkirakan dan data pada baris tersebut masih berguna. Sedangkan baris dengan menu kosong dihapus menggunakan dropna() karena kita tidak mengetahui jenis menu yang dijual, sehingga data tersebut kurang berguna untuk analisis berdasarkan menu.

```python
#Latihan 5

print(df.duplicated().sum()) # jumlah baris duplikat
df = df.drop_duplicates()
df['harga'] = df['harga'].astype(int) # memastikan tipe data harga adalah integer
print(df.dtypes)
```

Analisis 5 :

Setelah `drop_duplicates()`, jumlah baris dataset menjadi lebih sedikit karena data yang duplikat dihapus. Memastikan `dtypes` benar penting agar setiap data dapat diolah dan dianalisis dengan tepat, terutama untuk perhitungan seperti harga.


```python
#Latihan 6

laris = df[df['terjual'] > 20] # filtering
urut = df.sort_values(by='terjual', ascending=True) # sorting
df['total_pendapatan'] = df['harga'] * df['terjual'] # kolom turunan
ringkasan = df.groupby('menu')['total_pendapatan'].sum() # agregasi
print(ringkasan)
print(urut)
```

Analisis 6 :

Menu dengan total_pendapatan tertinggi adalah menu yang paling menguntungkan. Data ini membantu kantin menentukan menu yang perlu diprioritaskan dan ditambah stoknya.
